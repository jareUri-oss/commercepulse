```python
import pandas as pd 

order_items = pd.read_csv("../data/raw/olist_order_items_dataset.csv")
products = pd.read_csv("../data/raw/olist_products_dataset.csv")
product_name = pd.read_csv("../data/raw/product_category_name_translation.csv")
```

```python
date_cols = [
    "shipping_limit_date"
]

for col in date_cols:
    order_items[col] = pd.to_datetime(order_items[col])

order_items.dtypes
```

```python
#contamos cuantos registros/items tiene cada orden 
items_per_orders = order_items.groupby("order_id") ["order_item_id"].count()
items_per_orders.head(100)
```

```python
#conteo absoluto (numero de ordenes por tamano de carrito)
(items_per_orders.value_counts(normalize=True) * 100).round(2).head()
```

```python
(items_per_orders.mean().round(2))
```

```python
items_per_orders.median()
```

```python
(items_per_orders.describe().round(2))
```

```python
order_items[["price", "freight_value"]].agg(["mean", "median"]).round(2)
```

```python
total_price = order_items["price"].sum()
total_freight = order_items["freight_value"].sum()

freight_ratio = (total_freight/ total_price) * 100

freight_ratio.round(2)
```
