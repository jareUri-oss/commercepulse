```python
import pandas as pd

#Se cambio el tipo de dato a str debido a que no es una cantidad es una etiqueta 01001 = 1001 int eliminaria el 0
customers = pd.read_csv("../data/raw/olist_customers_dataset.csv", dtype={"customer_zip_code_prefix": str})
```

```python
#con MAX() identificamos el elemento con mayor caracteres 
customers["customer_zip_code_prefix"].str.len().max()
```

```python
#con min() identificados el elemento con menor cantidad de caracteres
customers["customer_zip_code_prefix"].str.len().min()
```

```python
#.values_counts hace un conteo de cada elemento uno de los elementos de la columna
customers["customer_zip_code_prefix"].str.len().value_counts()
```

```python
#Determinamos si existen duplicados en todas las filas, con keep='first' determinamos cuales son los duplicados
#Los resultados muestran que no existen filas duplicadas 
customers.duplicated().value_counts()
```

```python
#Verificamos que la columna 'customer_id' no cuenta con datos duplicados 
customers.duplicated(subset=["customer_id"], keep=False).value_counts()
```

```python
#Verificamos que la columna 'customer_unique_id' cuenta con datos duplicados
customers.duplicated(subset=["customer_unique_id"], keep='first').value_counts()
```
